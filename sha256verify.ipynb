```python
%%bash -s {dataset}
DATA=$1
mkdir -p _sha256verify
af3io compress "$DATA" "_sha256verify/$DATA.af3io.sqfs"
(cd _sha256verify/ && af3io decompress "$DATA.af3io.sqfs" .)
(cd _sha256verify/ && ../sha256path "$DATA" > "$DATA.sha256.txt")
./sha256path "$DATA" > "$DATA.sha256.txt"
sha256sum "_sha256verify/$DATA.sha256.txt"
sha256sum "$DATA.sha256.txt"
```
